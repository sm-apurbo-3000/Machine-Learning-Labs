```python
#matrix properties
arr1 = np.array([[10, 20, 30], [40, 50, 60]])
arr2 = np.array([[40, 70, 80], [20, 50, 70]])

print(arr1 + arr2)
print(arr2 - arr1)
```

```python
arr1 = np.array([[10, 20, 30], [40, 50, 60]])
print(arr1.shape)
arr2 = np.array([[40, 70, 80], [20, 50, 70]])
print(arr2.shape)

print(arr1 * arr2) #dot product
#print(arr1 @ arr2)
```

```python
arr1 = np.array([[10, 20, 30], [40, 50, 60]]) #(2,3)
arr2 = np.array([[40, 70],[20, 50],[80, 70]]) #(3,2)

print(arr1 @ arr2) #cross product
```

```python
import matplotlib.pyplot as plt
```

```python
x = np.array([0,6])
y = np.array([10, 40])

plt.plot(x,y)
plt.show()
```

```python
points = [3, 5, 90, 23, 45]
plt.plot(points, marker="*", linestyle = "dotted", color="r")
plt.show()

x = np.arange(-10, 11, 1)
y = x ** 2
plt.title(" y = x * x")
plt.xlabel("values of x")
plt.ylabel("y er values")
plt.grid()
plt.plot(x,y)
plt.show()
```

```python
x = np.arange(-10, 11, 1)
y = x ** 2
plt.title(" y = x * x")
plt.xlabel("values of x")
plt.ylabel("y er values")
plt.grid()
plt.plot(x,y, color="orange", marker="*")
plt.show()
```

```python
x = np.arange(-10, 11, 1)

y1 = x ** 2
plt.subplot(3,2,1) #(total_row, total_col, serial)
plt.plot(x, y1)

y2 = x ** 3
plt.subplot(3,2,2) #(total_row, total_col, serial)
plt.plot(x, y2)

y3 = 2 * x
plt.subplot(3,2,3) #(total_row, total_col, serial)
plt.plot(x, y3)

y4 = np.sin(x)
plt.subplot(3,2,4) #(total_row, total_col, serial)
plt.plot(x, y4)

y5 = x - 2
plt.subplot(3,2,5) #(total_row, total_col, serial)
plt.plot(x, y5)

y6 = 3*x + 2
plt.subplot(3,2,6) #(total_row, total_col, serial)
plt.plot(x, y6)

plt.show()
```
