###Numpy Tasks

```python
import numpy as np
```

Task 1: Create two linear arrays a and b with 20 random elements each ranging from 1 to 100.

```python
a, b = np.random.randint(1, 101, 20), np.random.randint(1, 101, 20)

#print(a.shape, b.shape)
print(a)
print(b)
```

Task 2: Now change those linear arrays to two 5x4 matrices.

```python
a, b = a.reshape(5, 4), b.reshape(5, 4)

print(a)
print(b)
```

Task 3: Perform scalar multiplication with matrix a and b.

```python
scaler_multiplier = 5 # example scaler multiplier

a_scaled, b_scaled = a * scaler_multiplier, b * scaler_multiplier

print(a_scaled)
print(b_scaled)
```

Task 4: Now try to modify the matrix b and perform matrix multiplication with a and b to store it in
c.

```python
# For matrix multiplication, b must have a shape of (4, 5) to perform a * b. Reshaping b to (4, 5)...

b_modified = b.reshape(4, 5)

c = np.matmul(a, b_modified)

print(c)
```

Task 5: Print only the column 2, 3 of matrix c. Print the whole matrix using slicing technique.

```python
print('Column 2, 3 of matrix c:\n')
print(c[:, 1:3])

print('\nWhole matrix of c:\n')
print(c[:, :])
```

Task 6: Find the largest element in the array, and also the index of that element.

```python
max_val = c.max()
max_val_idx = np.unravel_index(np.argmax(c), c.shape)
x, y = max_val_idx[0], max_val_idx[1]
print(f'Largest element: {max_val} at index: ({x}, {y})')
```

Task 7: Convert the matrix c into an one dimensional array keeping the original array
unchanged.

```python
c_flattened = c.flatten()
print(f'Flattened array: \n{c_flattened}')
print(f'\nOriginal array: \n{c}')
```

###Matplotlib Tasks

```python
import matplotlib.pyplot as plt
```

Data for plots

```python
temperature_dhaka = [25, 34, 21, 45, 28, 6, 43, 18, 7, 2]
humidity_dhaka = [28, 25, 29, 20, 26, 50, 19, 29, 52, 55]

study_hours = [2, 3, 4, 4, 5, 6, 7, 7, 8, 9, 9, 10, 11, 11, 12]
marks = [6, 10, 15, 20, 34, 44, 55, 60, 55, 67, 70, 80, 90, 99, 100]

subjects = ['Maths', 'English', 'Science', 'Physics', 'Computer']
subject_marks = [89, 90, 45, 78, 99]
```

Task 1: Generate a scatter plot graph for temperature_dhaka vs humidity_dhaka using (*) as the
marker in color red.

```python
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.xlabel('Temperature in Celcius')
plt.ylabel('Humidity in Percentage')
plt.title('Temperature vs Humidity Scatter Plot in Dhaka')
plt.grid()
plt.show()
```

Task 2: Now do a line plot for study_hours vs marks with figure size (12, 8)

```python
plt.figure(figsize=(12, 8))

plt.plot(study_hours, marks, linewidth = 2)
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks Line Plot')
plt.grid()
plt.show()
```

Task 3: Now draw a horizontal bar graph showing the scores in each of the subjects.

```python
plt.barh(subjects, subject_marks, color='grey')
plt.xlabel('Subject Marks')
plt.ylabel('Subjects')
plt.title('Subject Marks (Horizontal Bar Graph)')
plt.grid()
plt.show()
```

Task 4: Draw a vertical bar graph where each subject score is visualized in different colors.

```python
plt.bar(subjects, subject_marks, color=['red', 'green', 'blue', 'orange', 'purple'])
plt.xlabel('Subjects')
plt.ylabel('Subject Marks')
plt.title('Subject Marks (Vertical Bar Graph)')
plt.grid()
plt.show()
```

Task 5: Show all these graphs in one graph using subplot( ) function.

```python
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
plt.scatter(temperature_dhaka, humidity_dhaka, color='red', marker='*')
plt.xlabel('Temperature in Celcius')
plt.ylabel('Humidity in Percentage')
plt.title('Temperature vs Humidity Scatter Plot in Dhaka')
plt.grid()

plt.subplot(2, 2, 2)
plt.plot(study_hours, marks, linewidth = 2)
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks Line Plot')
plt.grid()

plt.subplot(2, 2, 3)
plt.barh(subjects, subject_marks, color='grey')
plt.xlabel('Subject Marks')
plt.ylabel('Subjects')
plt.title('Subject Marks (Horizontal Bar Graph)')
plt.grid()

plt.subplot(2, 2, 4)
plt.bar(subjects, subject_marks, color=['red', 'green', 'blue', 'orange', 'purple'])
plt.xlabel('Subjects')
plt.ylabel('Subject Marks')
plt.title('Subject Marks (Vertical Bar Graph)')
plt.grid()

plt.tight_layout()
plt.show()
```
